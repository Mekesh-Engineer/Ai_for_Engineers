# Experiment 7: Text Summarization using Large Language Models (LLM)
## Notebook 1: Dataset Exploration, Preprocessing & Corpus Analytics

This notebook explores the multi-domain document corpus, analyzes document and summary word count distributions, tokenization characteristics, and verifies data integrity.

```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Ensure project root and src/ are importable
PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))
sys.path.insert(0, os.path.join(PROJECT_ROOT, "src"))

from data_loader import get_or_create_raw_articles, split_dataset
from data_preprocessing import preprocess_corpus, verify_data_integrity, compute_text_statistics
```

### 1. Load Raw Articles Corpus

```python
raw_csv_path = os.path.join(PROJECT_ROOT, "data", "raw", "articles.csv")
df_raw = get_or_create_raw_articles(raw_csv_path)
print(f"Total raw documents loaded: {len(df_raw)}")
df_raw.head(3)
```

### 2. Preprocess & Clean Corpus

```python
df_processed = preprocess_corpus(df_raw, min_words=50, max_words=3000)
integrity = verify_data_integrity(df_processed)
print("Integrity Verification:", integrity)
df_processed[["id", "category", "title", "doc_word_count", "ref_word_count"]]
```

### 3. Length Distribution Analysis (Documents vs Reference Summaries)

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4), dpi=150)

ax1.hist(df_processed["doc_word_count"], bins=10, color="#2b5c8f", alpha=0.8, edgecolor="black")
ax1.set_title("Document Word Count Distribution", fontweight="bold")
ax1.set_xlabel("Word Count")
ax1.set_ylabel("Frequency")

ax2.hist(df_processed["ref_word_count"], bins=10, color="#e27c3e", alpha=0.8, edgecolor="black")
ax2.set_title("Reference Summary Word Count Distribution", fontweight="bold")
ax2.set_xlabel("Word Count")
ax2.set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

### 4. Corpus Category Breakdown & Sentence Statistics

```python
for idx, row in df_processed.iterrows():
    stats = compute_text_statistics(row["document_text"])
    print(f"[{row['id']}] {row['title'][:45]}...")
    print(f"   Category: {row['category']}")
    print(f"   Words: {stats['word_count']} | Sentences: {stats['sentence_count']} | Avg Sent Length: {stats['avg_sentence_length_words']} words")
    print("-" * 70)
```
