```python
import os
import sys
import pandas as pd
import matplotlib.pyplot as plt

# Ensure src is importable
sys.path.insert(0, os.path.abspath('..'))
from src.data_loader import DataLoaderModule
from src.data_preprocessing import PreprocessingModule
```
